#### Filling Missing Values with Mean and Mode

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'Age':[25,30,np.nan,40,35],'Department':['HR','Finance','Finance',np.nan,'IT']})
print("Original Dataset(with missing values):")
print(df)
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
print("\nDataset after filling missing values:")
print(df)
```

#### Forward Fill and Backward Fill

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({'Age':[25,30,np.nan,40,35],'Department':['HR','Finance','Finance',np.nan,'IT']})
print("Original Dataset(with missing values):")
print(df)
df_ffill = df.copy()
df_ffill.ffill(inplace=True)
print(df_ffill)
df_bfill = df.copy()
df_bfill.bfill(inplace=True)
print(df_bfill)
```

#### Removing Rows and Columns with Missing Values

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({ 'Age': [25,30,np.nan,40,35],
     'Department': ['HR','Finance','Finance',np.nan,'IT']})
print('Orginal Dataset(With Missing Values):')
print(df)
df_drop = df.copy()
df_drop.dropna(inplace=True)
print(df_drop)
df_drop_col = df.copy()
df_drop_col.dropna(axis=1, inplace=True)
print(df_drop_col)
```

#### Removing Duplicate Records

```python
import pandas as pd
df = pd.DataFrame({
    'ID': [1, 2, 2, 3, 4, 5],
    'Name': ['John', 'Jane', 'Jane', 'Mike', 'Emily', 'David'],
    'Age': [25, 30, 30, 35, 28, 40]
})
print('Original Dataset:')
print(df)
df_exact = df.drop_duplicates(subset='ID')
print('Dataset after removing duplicates:')
print(df_exact)
```

#### Standardizing Date Formats

```python
import pandas as pd
df = pd.DataFrame({
    'Date': ['2025-01-05', '05/01/2025', 'Jan 5, 2025', '20250105', '2025.01.05'],
})
print('Original Dataset:')
print(df)
df['Date'] = pd.to_datetime(df['Date'],errors = 'coerce').dt.strftime('%y-%m-%d')
print(df)
```

#### Changing Text Case

```python
df = pd.DataFrame({
    'Name':['Alice','BOB','Charlie','DAVID']
})
print("Original DataFrame: ")
print(df)
df['Name_lower'] = df['Name'].str.lower()
df['Name_upper'] = df['Name'].str.upper()
df['Name_title'] = df['Name'].str.title()
print(df)
```
